```python
import numpy as np
import pandas as pd

# 1. Initialize Random Seed (use your roll number)
np.random.seed(65)   # 🔁 replace 123 with your roll number

# 2. Generate Sales Data (12 months × 4 categories)
sales_data = np.random.randint(1000, 5000, size=(12, 4))

# 3. Convert to DataFrame
months = ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"]
columns = ["Electronics", "Clothing", "Home & Kitchen", "Sports"]

df = pd.DataFrame(sales_data, columns=columns, index=months)
df
```

```python
df.head()
df.describe()
```

```python
# Total per category
category_total = df.sum()

# Total per month
monthly_total = df.sum(axis=1)

category_total, monthly_total
```

```python
growth = df.pct_change()
growth.mean()
```

```python
df["Total Sales"] = df.sum(axis=1)
df["Growth Rate"] = df["Total Sales"].pct_change() * 100
df
```

```python
roll_no = 65  # 🔁 replace with your roll number

if roll_no % 2 == 0:
    df["Electronics"] = df["Electronics"] * 0.9
else:
    df["Clothing"] = df["Clothing"] * 0.85

df
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df[["Electronics","Clothing","Home & Kitchen","Sports"]].plot(marker='o')
plt.title("Monthly Sales Trends")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
sns.boxplot(data=df[["Electronics","Clothing","Home & Kitchen","Sports"]])
plt.title("Sales Distribution")
plt.show()
```

```python
array = np.array([[1, -2, 3],[-4, 5, -6]])

# i. Absolute value
np.abs(array)
```

```python
# ii. Percentiles
flat = array.flatten()

np.percentile(flat, [25, 50, 75])          # flattened
np.percentile(array, [25,50,75], axis=0)   # column-wise
np.percentile(array, [25,50,75], axis=1)   # row-wise
```

```python
# iii. Mean, Median, Std
np.mean(flat), np.median(flat), np.std(flat)

np.mean(array, axis=0)   # column
np.mean(array, axis=1)   # row
```
